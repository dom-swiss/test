```python
# Prompt
'''
Write a Python code that identifies the columns with missing values in a pandas data frame.
'''
# Function to find columns with missing values
def find_missing_columns(df):
    # Use isnull().sum() to find number of missing values in each column
    # Columns with any missing values will have a sum > 0
    # Filter and return these columns
    missing_columns = df.columns[df.isnull().sum() > 0]
    return missing_columns

# Call the function
missing_cols = find_missing_columns(df)

# Print the columns with missing values
print("Columns with missing values: ", missing_cols)
```

```python
# Prompt
'''
Write a Python code to replace the missing values in a pandas data frame, per the following guidelines.
1. For a categorical attribute "Screen_Size_cm", replace the missing values with the most frequent value in the column.
2. For a continuous value attribute "Weight_kg", replace the missing values with the mean value of the entries in the column.
'''

import numpy as np
# Function to impute missing values
def impute_missing_values(df):
    for col in df.columns:
        if df[col].dtype == 'object':  # Categorical column
            mode_value = df[col].mode()[0] if not df[col].mode().empty else None
            if mode_value is not None:
                df[col].fillna(mode_value, inplace=True)
        elif df[col].dtype in ['int64', 'float64']:  # Continuous column
            mean_value = df[col].mean()
            df[col].fillna(mean_value, inplace=True)
    return df

# Define columns to handle
specific_cols = ['Screen_Size_cm', 'Weight_kg']

# Apply imputation
imputed_df = impute_missing_values(df[specific_cols])

# Display the result
print(imputed_df)
```

```python
# Prompt
'''
Write a Python code to normalize the content under the attribute "CPU_frequency" in a data frame df concerning its maximum value. Make changes to the original data, and do not create a new attribute
'''
# Function to normalize 'CPU_frequency' column
def normalize_cpu_frequency(df):
    # Find the maximum value of 'CPU_frequency'
    max_cpu = df['CPU_frequency'].max()
    
    # Normalize 'CPU_frequency' by dividing by max_cpu
    df['CPU_frequency'] = df['CPU_frequency'].apply(lambda x: x / max_cpu if not pd.isna(x) else np.nan)
    return df

# Apply the function
df_normalized = normalize_cpu_frequency(df)

# Display the result
print(df_normalized)
```

```python
# Prompt
'''
Write a Python code to perform the following tasks.
1. Convert a data frame df attribute "Screen", into indicator variables, saved as df1, with the naming convention "Screen_<unique value of the attribute>".
2. Append df1 into the original data frame df.
3. Drop the original attribute from the data frame df.
'''

# Function to convert 'Screen' to indicator variables
def convert_to_indicators(df2):
    # Use get_dummies to create indicator variables
    df1 = pd.get_dummies(df2, columns=['Screen'], prefix='Screen', prefix_sep='_')
    
    return df1

# Apply the function
df1 = convert_to_indicators(df_normalized)

# Append df1 to the original df
df = pd.concat([df_normalized, df1], axis=1)

# Drop the original 'Screen' column
df.drop(columns=['Screen'], inplace=True)

# Display the resulting DataFrame
print(df)
```
